```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_regression
```

```python
X, y = make_regression(n_samples=4, n_features=1, n_informative=1, n_targets=1, noise=80, random_state=13)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y)
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(X, y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(X, y)
plt.plot(X, reg.predict(X), color="purple")
```

```python
# Assume slope to be constant m = 78.35
y_pred = ((78.35*X) + 0).reshape(4)
```

```python
plt.scatter(X, y)
plt.plot(X, reg.predict(X), color="purple")
plt.plot(X, y_pred, color="brown")
```

```python
m = 78.35
b = 0

loss_slope = -2 * np.sum(y-m*X.ravel() - b)
loss_slope
```

```python
# Lets take lr as 0.1
lr = 0.1
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
```

```python
b
```

```python
b = -100
m = 78.35
lr = 0.1

epoch = 10

for i in range(epoch):
    slope = -2 * np.sum(y - m*X.ravel() - b)

    b = b - (lr*slope)
    y_pred = m*X + b
    plt.plot(X, y_pred)

plt.scatter(X, y, label="X")

```
